````python
import pandas as pd
import numpy as np

# =========================================================
# ASHLEY BRANCH - DATA PREPARATION
# =========================================================
# STEP 1 - LOAD DATA
# =========================================================

# Change these to your actual file names

sale = pd.read_csv("sale.csv")
purchase = pd.read_csv("purchase.csv")
inventory = pd.read_csv("inventory.csv")


# STEP 2 - CLEAN COLUMN NAMES
# =========================================================

sale.columns = sale.columns.str.strip()
purchase.columns = purchase.columns.str.strip()
inventory.columns = inventory.columns.str.strip()



# =========================================================
# STEP 3 - CLEAN BRANCH NAMES
# =========================================================

for data in [sale, purchase, inventory]:

    data["Branch"] = (
        data["Branch"]
        .astype(str)
        .str.strip()
    )


# =========================================================
# STEP 4 - KEEP ONLY ASHLEY BRANCH
# =========================================================

sale = sale[
    sale["Branch"].str.lower() == "ashley"
].copy()

purchase = purchase[
    purchase["Branch"].str.lower() == "ashley"
].copy()

inventory = inventory[
    inventory["Branch"].str.lower() == "ashley"
].copy()




print("Ashley branch selected")

print("Sale rows:", len(sale))
print("Purchase rows:", len(purchase))
print("Inventory rows:", len(inventory))



# =========================================================
# STEP 5 - CONVERT DATES
# =========================================================

sale["Date"] = pd.to_datetime(
    sale["Date"],
    errors="coerce"
)

purchase["Date"] = pd.to_datetime(
    purchase["Date"],
    errors="coerce"
)

inventory["Last Updated"] = pd.to_datetime(
    inventory["Last Updated"],
    errors="coerce"
)



# =========================================================
# STEP 6 - CLEAN STOCK CODES
# =========================================================

for data in [sale, purchase, inventory]:

    data["Stock Code"] = (
        data["Stock Code"]
        .astype(str)
        .str.strip()
        .str.upper()
    )


# =========================================================
# STEP 7 - CLEAN DESCRIPTION
# =========================================================

for data in [sale, purchase, inventory]:

    if "Description" in data.columns:

        data["Description"] = (
            data["Description"]
            .astype(str)
            .str.strip()
        )


# =========================================================
# STEP 8 - CONVERT SALE NUMBERS
# =========================================================

sale_numeric = [
    "Selling Price",
    "Qty",
    "Net Amount",
    "Buying Price",
    "Profit",
    "Profit Margin %"
]

for col in sale_numeric:

    if col in sale.columns:

        sale[col] = pd.to_numeric(
            sale[col],
            errors="coerce"
        )


# =========================================================
# STEP 9 - CONVERT PURCHASE NUMBERS
# =========================================================

purchase_numeric = [
    "Quantity",
    "Buying Price"
]

for col in purchase_numeric:

    if col in purchase.columns:

        purchase[col] = pd.to_numeric(
            purchase[col],
            errors="coerce"
        )


# =========================================================
# STEP 10 - CONVERT INVENTORY NUMBERS
# =========================================================

inventory_numeric = [
    "On Hand Qty",
    "Buying Price",
    "Selling Price"
]

for col in inventory_numeric:

    if col in inventory.columns:

        inventory[col] = pd.to_numeric(
            inventory[col],
            errors="coerce"
        )


# =========================================================
# STEP 11 - DATE RANGE
# =========================================================

print("\n========== DATE RANGE ==========")

print("\nSALE")
print("Start:", sale["Date"].min())
print("End:", sale["Date"].max())

print("\nPURCHASE")
print("Start:", purchase["Date"].min())
print("End:", purchase["Date"].max())

print("\nINVENTORY")
print("Start:", inventory["Last Updated"].min())
print("End:", inventory["Last Updated"].max())


# =========================================================
# STEP 12 - MISSING VALUES
# =========================================================

print("\n========== MISSING VALUES ==========")

print("\nSALE")
print(
    sale[
        ["Date", "Stock Code", "Qty", "Net Amount"]
    ].isna().sum()
)

print("\nPURCHASE")
print(
    purchase[
        ["Date", "Stock Code", "Quantity", "Buying Price"]
    ].isna().sum()
)

print("\nINVENTORY")
print(
    inventory[
        ["Last Updated", "Stock Code", "On Hand Qty"]
    ].isna().sum()
)


# =========================================================
# STEP 13 - CHECK DATA TYPES
# =========================================================

print("\n========== DATA TYPES ==========")

print("\nSALE")
print(sale.dtypes)

print("\nPURCHASE")
print(purchase.dtypes)

print("\nINVENTORY")
print(inventory.dtypes)


# =========================================================
# STEP 14 - PREVIEW DATA
# =========================================================

print("\n========== SALE SAMPLE ==========")

print(
    sale[
        [
            "Date",
            "Stock Code",
            "Description",
            "Qty",
            "Selling Price",
            "Buying Price",
            "Profit"
        ]
    ].head()
)


print("\n========== PURCHASE SAMPLE ==========")

print(
    purchase[
        [
            "Date",
            "Stock Code",
            "Description",
            "Quantity",
            "Buying Price"
        ]
    ].head()
)


print("\n========== INVENTORY SAMPLE ==========")

print(
    inventory[
        [
            "Last Updated",
            "Stock Code",
            "Description",
            "Group",
            "On Hand Qty",
            "Buying Price",
            "Selling Price"
        ]
    ].head()
)


print("\n====================================")
print("ASHLEY DATA PREPARATION COMPLETE")
print("====================================")
````

Step2 - Create a Product Master. Build one row per Stock Code containing fields such as Description, Group/Category, current Selling Price, current Buying Price, and current On Hand Qty. This becomes the main reference table for every later layer.



````python
# =========================================================
# STEP 2 - CREATE PRODUCT MASTER
# ASHLEY BRANCH ONLY
# =========================================================


# ---------------------------------------------------------
# 2.0 FILTER ALL DATA TO ASHLEY ONLY
# ---------------------------------------------------------

sale = sale[
    sale["Branch"].astype(str).str.strip().str.lower() == "ashley"
].copy()

purchase = purchase[
    purchase["Branch"].astype(str).str.strip().str.lower() == "ashley"
].copy()

inventory = inventory[
    inventory["Branch"].astype(str).str.strip().str.lower() == "ashley"
].copy()


# Check branch before continuing
print("========== ASHLEY FILTER CHECK ==========")

print("Sale branches:")
print(sale["Branch"].value_counts())

print("\nPurchase branches:")
print(purchase["Branch"].value_counts())

print("\nInventory branches:")
print(inventory["Branch"].value_counts())


# ---------------------------------------------------------
# 2.1 CREATE LIST OF EVERY PRODUCT
# ---------------------------------------------------------

all_codes = (
    set(sale["Stock Code"])
    | set(inventory["Stock Code"])
    | set(purchase["Stock Code"])
)

product_master = pd.DataFrame({
    "Stock Code": sorted(all_codes)
})

# Add Branch
product_master["Branch"] = "Ashley"


# ---------------------------------------------------------
# 2.2 PREPARE INVENTORY INFORMATION
# ---------------------------------------------------------

inventory_info = (
    inventory[
        [
            "Stock Code",
            "Description",
            "Group",
            "Selling Price",
            "Buying Price",
            "On Hand Qty"
        ]
    ]
    .drop_duplicates(
        subset="Stock Code",
        keep="last"
    )
)

inventory_info = inventory_info.rename(
    columns={
        "Description": "Description_Inventory"
    }
)


# ---------------------------------------------------------
# 2.3 PREPARE SALE INFORMATION
# ---------------------------------------------------------

# SALE does NOT contain Group
# so we use only Stock Code + Description here

sale_info = (
    sale[
        [
            "Stock Code",
            "Description"
        ]
    ]
    .drop_duplicates(
        subset="Stock Code",
        keep="last"
    )
)

sale_info = sale_info.rename(
    columns={
        "Description": "Description_Sale"
    }
)


# ---------------------------------------------------------
# 2.4 MERGE INVENTORY INFORMATION
# ---------------------------------------------------------

product_master = product_master.merge(
    inventory_info,
    on="Stock Code",
    how="left"
)


# ---------------------------------------------------------
# 2.5 MERGE SALE INFORMATION
# ---------------------------------------------------------

product_master = product_master.merge(
    sale_info,
    on="Stock Code",
    how="left"
)


# ---------------------------------------------------------
# 2.6 CREATE FINAL DESCRIPTION
# ---------------------------------------------------------

# First use Inventory description.
# If missing, use Sale description.

product_master["Description"] = (
    product_master["Description_Inventory"]
    .fillna(product_master["Description_Sale"])
)


# ---------------------------------------------------------
# 2.7 REMOVE TEMPORARY DESCRIPTION COLUMNS
# ---------------------------------------------------------

product_master = product_master.drop(
    columns=[
        "Description_Inventory",
        "Description_Sale"
    ]
)


# ---------------------------------------------------------
# 2.8 ADD SOURCE FLAGS
# ---------------------------------------------------------

product_master["In Sale"] = (
    product_master["Stock Code"]
    .isin(sale["Stock Code"])
)

product_master["In Inventory"] = (
    product_master["Stock Code"]
    .isin(inventory["Stock Code"])
)

product_master["In Purchase"] = (
    product_master["Stock Code"]
    .isin(purchase["Stock Code"])
)


# ---------------------------------------------------------
# 2.9 REORDER COLUMNS
# ---------------------------------------------------------

product_master = product_master[
    [
        "Branch",
        "Stock Code",
        "Description",
        "Group",
        "Selling Price",
        "Buying Price",
        "On Hand Qty",
        "In Sale",
        "In Inventory",
        "In Purchase"
    ]
]


# ---------------------------------------------------------
# 2.10 CHECK RESULT
# ---------------------------------------------------------

print("\n========== PRODUCT MASTER ==========")

print("\nBranch:")
print(product_master["Branch"].value_counts())

print("\nTotal Products:")
print(len(product_master))

print("\nMissing Values:")
print(product_master.isna().sum())

print("\nSample:")
print(product_master.head(20).to_string(index=False))
````

````python
# =========================================================
# STEP 2 DIAGNOSTIC ONLY
# CHECK WHEN THE 677 MISSING PRODUCTS WERE LAST SOLD
# DO NOT MODIFY PRODUCT MASTER
# =========================================================

import pandas as pd

# Reference date = Purchase / Inventory snapshot date
reference_date = pd.Timestamp("2026-09-04")

# ---------------------------------------------------------
# 1. Get the 677 Stock Codes
# ---------------------------------------------------------

missing_codes = set(
    product_master.loc[
        product_master["Selling Price"].isna()
        & product_master["Buying Price"].isna()
        & product_master["On Hand Qty"].isna(),
        "Stock Code"
    ]
)

# ---------------------------------------------------------
# 2. Filter Sale data for only those missing products
# ---------------------------------------------------------

missing_sales = sale[
    sale["Stock Code"].isin(missing_codes)
].copy()

missing_sales["Date"] = pd.to_datetime(
    missing_sales["Date"],
    errors="coerce"
)

# ---------------------------------------------------------
# 3. Find First Sale Date and Last Sale Date per product
# ---------------------------------------------------------

missing_product_dates = (
    missing_sales
    .groupby("Stock Code")["Date"]
    .agg(
        First_Sale_Date="min",
        Last_Sale_Date="max"
    )
    .reset_index()
)

# ---------------------------------------------------------
# 4. Calculate Days Since Last Sale
# ---------------------------------------------------------

missing_product_dates["Days_Since_Last_Sale"] = (
    reference_date - missing_product_dates["Last_Sale_Date"]
).dt.days

# ---------------------------------------------------------
# 5. Overall summary
# ---------------------------------------------------------

total_missing = len(missing_product_dates)

earliest_first_sale = missing_product_dates["First_Sale_Date"].min()
latest_last_sale = missing_product_dates["Last_Sale_Date"].max()

# ---------------------------------------------------------
# 6. Count products by how long ago they were last sold
# ---------------------------------------------------------

sold_last_30_days = (
    missing_product_dates["Days_Since_Last_Sale"] <= 30
).sum()

sold_31_90_days = (
    missing_product_dates["Days_Since_Last_Sale"].between(31, 90)
).sum()

sold_91_180_days = (
    missing_product_dates["Days_Since_Last_Sale"].between(91, 180)
).sum()

sold_181_365_days = (
    missing_product_dates["Days_Since_Last_Sale"].between(181, 365)
).sum()

sold_over_365_days = (
    missing_product_dates["Days_Since_Last_Sale"] > 365
).sum()

# ---------------------------------------------------------
# 7. Print simple summary
# ---------------------------------------------------------

print("========== 677 MISSING PRODUCT TIME SUMMARY ==========")

print(f"\nMissing products: {total_missing}")

print(f"\nEarliest first sale: {earliest_first_sale.date()}")
print(f"Latest last sale:    {latest_last_sale.date()}")

print("\nLast Sale Period:")
print(f"Within last 30 days:    {sold_last_30_days}")
print(f"31 - 90 days ago:       {sold_31_90_days}")
print(f"91 - 180 days ago:      {sold_91_180_days}")
print(f"181 - 365 days ago:     {sold_181_365_days}")
print(f"More than 365 days ago: {sold_over_365_days}")

# ---------------------------------------------------------
# 8. Optional detailed table
# ---------------------------------------------------------

missing_product_dates = missing_product_dates.sort_values(
    "Last_Sale_Date",
    ascending=False
)

print("\n========== PRODUCT DETAIL ==========")

print(
    missing_product_dates[
        [
            "Stock Code",
            "First_Sale_Date",
            "Last_Sale_Date",
            "Days_Since_Last_Sale"
        ]
    ].to_string(index=False)
)
````

````python
# =========================================================
# STEP 2.1 DIAGNOSTIC ONLY
# FIND WHICH DATA SOURCE THE 24 PRODUCTS ARE MISSING FROM
# DO NOT MODIFY PRODUCT MASTER
# =========================================================

import pandas as pd

# ---------------------------------------------------------
# 1. Get the 24 products sold within the last 30 days
# ---------------------------------------------------------

recent_24 = missing_product_dates[
    missing_product_dates["Days_Since_Last_Sale"] <= 30
].copy()

recent_24_codes = set(recent_24["Stock Code"])

print("Recent products to investigate:", len(recent_24_codes))


# ---------------------------------------------------------
# 2. Get Stock Codes from each source
# ---------------------------------------------------------

sale_codes = set(
    sale["Stock Code"].dropna().astype(str).str.strip()
)

purchase_codes = set(
    purchase["Stock Code"].dropna().astype(str).str.strip()
)

inventory_codes = set(
    inventory["Stock Code"].dropna().astype(str).str.strip()
)


# ---------------------------------------------------------
# 3. Check each of the 24 products
# ---------------------------------------------------------

check_24 = recent_24[
    ["Stock Code", "Last_Sale_Date", "Days_Since_Last_Sale"]
].copy()

check_24["Stock Code"] = (
    check_24["Stock Code"]
    .astype(str)
    .str.strip()
)

check_24["In Sales"] = check_24["Stock Code"].isin(sale_codes)

check_24["In Purchase"] = check_24["Stock Code"].isin(purchase_codes)

check_24["In Inventory"] = check_24["Stock Code"].isin(inventory_codes)


# ---------------------------------------------------------
# 4. Identify WHICH SOURCE is missing
# ---------------------------------------------------------

def identify_missing_source(row):

    missing = []

    if not row["In Purchase"]:
        missing.append("Purchase")

    if not row["In Inventory"]:
        missing.append("Inventory")

    if len(missing) == 0:
        return "None"

    return " + ".join(missing)


check_24["Missing From"] = check_24.apply(
    identify_missing_source,
    axis=1
)


# ---------------------------------------------------------
# 5. Show result for all 24 products
# ---------------------------------------------------------

print("\n========== 24 RECENT PRODUCTS ==========")

print(
    check_24[
        [
            "Stock Code",
            "Last_Sale_Date",
            "In Sales",
            "In Purchase",
            "In Inventory",
            "Missing From"
        ]
    ].to_string(index=False)
)


# ---------------------------------------------------------
# 6. Summary
# ---------------------------------------------------------

print("\n========== MISSING SOURCE SUMMARY ==========")

print(check_24["Missing From"].value_counts())


# ---------------------------------------------------------
# 7. Separate suspicious groups
# ---------------------------------------------------------

missing_purchase_only = check_24[
    (check_24["In Purchase"] == False) &
    (check_24["In Inventory"] == True)
]

missing_inventory_only = check_24[
    (check_24["In Purchase"] == True) &
    (check_24["In Inventory"] == False)
]

missing_both = check_24[
    (check_24["In Purchase"] == False) &
    (check_24["In Inventory"] == False)
]


print("\nMissing from PURCHASE only:")
print(len(missing_purchase_only))

print("\nMissing from INVENTORY only:")
print(len(missing_inventory_only))

print("\nMissing from BOTH Purchase + Inventory:")
print(len(missing_both))
````

````python
# =========================================================
# SIMPLE MISSING DESCRIPTION REPORT
# Source + Branch + Date + Missing Count
# =========================================================

import pandas as pd


def missing_description_report(data, source_name, date_column):

    df = data.copy()

    # Convert date
    df[date_column] = pd.to_datetime(
        df[date_column],
        errors="coerce"
    )

    # Find missing Description
    missing = df[
        df["Description"].isna()
        | df["Description"].astype("string").str.strip().isin(["", "nan", "None", "null"])
    ].copy()

    # Keep only date
    missing["Date"] = missing[date_column].dt.date

    # Count missing rows by branch and date
    result = (
        missing
        .groupby(["Branch", "Date"], dropna=False)
        .size()
        .reset_index(name="Missing Description Count")
    )

    # Add source
    result.insert(0, "Data Source", source_name)

    return result
````

````python
# ---------------------------------------------------------
# 3.2 DESCRIPTION + GROUP RELATIONSHIPS
# FIND SAFE DESCRIPTION -> GROUP PATTERNS
# FILL MISSING GROUPS
# --------------------------------------------------------
# -FILL MISSING GROUP DATA
#
# SITUATION 1:
# Description has only ONE known Group
# -> SAFE -> Fill missing Group
#
# SITUATION 2:
# Description has MULTIPLE known Groups
# -> UNSAFE -> Keep <NA>
# ---------------------------------------------------------

import pandas as pd


# =========================================================
# STEP 1: COUNT BEFORE FILLING
# =========================================================

total_products = all_sales["Stock Code"].nunique()

missing_before = (
    all_sales.loc[
        all_sales["Group"].isna(),
        "Stock Code"
    ]
    .nunique()
)


# =========================================================
# STEP 2: CLEAN / REDUCE DESCRIPTION
#
# Purpose:
# Reduce formatting differences into ONE description pattern.
#
# Examples:
# IQ SOFT       -> IQSOFT
# IQ-SOFT       -> IQSOFT
# Golden Duck   -> GOLDENDUCK
#
# Myanmar characters are preserved.
# =========================================================

all_sales["Description_Clean"] = (
    all_sales["Description"]
    .astype("string")
    .str.upper()
    .str.strip()
    .str.replace(r"[^\w]", "", regex=True)
    .str.replace("_", "", regex=False)
)


# =========================================================
# STEP 3: USE ONLY EXISTING GROUP DATA TO LEARN PATTERNS
# =========================================================

known_group_data = (
    all_sales.loc[
        all_sales["Group"].notna()
        & all_sales["Description_Clean"].notna()
        & (all_sales["Description_Clean"] != ""),
        ["Description_Clean", "Group"]
    ]
    .drop_duplicates()
)


# =========================================================
# STEP 4: COUNT HOW MANY GROUPS EACH DESCRIPTION HAS
# =========================================================

description_group_count = (
    known_group_data
    .groupby("Description_Clean")["Group"]
    .nunique()
)


# =========================================================
# SITUATION 1: SAFE DESCRIPTION
#
# Only ONE known Group
#
# Example:
# BEEBEE -> LADY
#
# Therefore:
# BEEBEE + <NA> -> LADY
# =========================================================

safe_descriptions = (
    description_group_count[
        description_group_count == 1
    ]
    .index
)


safe_group_map = (
    known_group_data[
        known_group_data["Description_Clean"]
        .isin(safe_descriptions)
    ]
    .drop_duplicates("Description_Clean")
    .set_index("Description_Clean")["Group"]
)


# =========================================================
# SITUATION 2: UNSAFE DESCRIPTION
#
# More than ONE known Group
#
# Example:
# ABWALK -> MEN
# ABWALK -> LADY
#
# Therefore:
# ABWALK + <NA> -> KEEP <NA>
# =========================================================

unsafe_descriptions = (
    description_group_count[
        description_group_count > 1
    ]
    .index
)


# =========================================================
# STEP 5: FILL ONLY SAFE MISSING GROUPS
# =========================================================

missing_mask = all_sales["Group"].isna()

all_sales.loc[
    missing_mask,
    "Group"
] = (
    all_sales.loc[
        missing_mask,
        "Description_Clean"
    ]
    .map(safe_group_map)
)


# =========================================================
# STEP 6: COUNT AFTER FILLING
# =========================================================

missing_after = (
    all_sales.loc[
        all_sales["Group"].isna(),
        "Stock Code"
    ]
    .nunique()
)

group_generated = missing_before - missing_after


# =========================================================
# STEP 7: PRODUCT LEVEL RESULT
# =========================================================

print("\n========== MISSING GROUP FILLING RESULT ==========")

print(f"Total Products               : {total_products}")
print(f"Missing Before Filling       : {missing_before}")
print(f"Group Generated              : {group_generated}")
print(f"Missing After Filling        : {missing_after}")


# =========================================================
# STEP 8: DESCRIPTION PATTERN RESULT
# =========================================================

total_descriptions = all_sales["Description_Clean"].nunique()

safe_description_count = len(safe_descriptions)
unsafe_description_count = len(unsafe_descriptions)


print("\n========== DESCRIPTION GROUP PATTERN ==========")

print(f"Total Clean Descriptions     : {total_descriptions}")
print(f"Safe Single-Group Pattern    : {safe_description_count}")
print(f"Unsafe Multi-Group Pattern   : {unsafe_description_count}")


# =========================================================
# STEP 9: SHOW SAFE DESCRIPTION -> GROUP RULES
# =========================================================

safe_pattern_table = (
    safe_group_map
    .reset_index()
    .rename(columns={"Group": "Generated_Group"})
    .sort_values("Description_Clean")
)


print("\n========== SAFE DESCRIPTION -> GROUP RULE ==========")

print(
    safe_pattern_table
    .to_string(index=False)
)


# =========================================================
# STEP 10: SHOW UNSAFE DESCRIPTION PATTERNS
# =========================================================

unsafe_pattern_table = (
    known_group_data[
        known_group_data["Description_Clean"]
        .isin(unsafe_descriptions)
    ]
    .groupby("Description_Clean")["Group"]
    .apply(lambda x: ", ".join(sorted(x.unique())))
    .reset_index(name="Possible_Groups")
    .sort_values("Description_Clean")
)


print("\n========== UNSAFE DESCRIPTION -> MULTIPLE GROUPS ==========")

print(
    unsafe_pattern_table
    .to_string(index=False)
)


# =========================================================
# STEP 11: FINAL DESCRIPTION + GROUP SUMMARY
# =========================================================

description_summary = (
    all_sales[
        ["Stock Code", "Description",
         "Description_Clean", "Group"]
    ]
    .drop_duplicates()
    .groupby(
        ["Description", "Description_Clean", "Group"],
        dropna=False
    )
    .agg(
        Product_Count=("Stock Code", "nunique")
    )
    .reset_index()
    .sort_values(
        ["Description_Clean", "Group", "Product_Count"],
        ascending=[True, True, False]
    )
)


print("\n========== DESCRIPTION + GROUP SUMMARY AFTER FILLING ==========")

print(
    description_summary
    .to_string(index=False)
)
````

Step 5.3 — Create Stock Coverage Status
For this first decision-support version, use these thresholds:
Out of Stock: 0 stock
Low Coverage: less than 1 month
Adequate: 1–3 months
High Coverage: more than 3 months
No Sales: N-class product

````python
# ==========================================================
# STEP 5.3 — CREATE STOCK COVERAGE STATUS
# ==========================================================

def stock_status(row):

    # No historical sales
    if row["ABC_Class"] == "N":
        return "No Sales"

    # No current stock
    elif row["On Hand Qty"] <= 0:
        return "Out of Stock"

    # Less than 1 month stock
    elif row["Stock_Coverage_Months"] < 1:
        return "Low Coverage"

    # 1 to 3 months stock
    elif row["Stock_Coverage_Months"] <= 3:
        return "Adequate"

    # More than 3 months stock
    else:
        return "High Coverage"


product_master["Stock_Status"] = product_master.apply(
    stock_status,
    axis=1
)


# ----------------------------------------------------------
# CHECK RESULT
# ----------------------------------------------------------

print("========== STOCK STATUS ==========")

print(
    product_master["Stock_Status"]
    .value_counts()
)

print("\n========== STOCK STATUS BY ABC CLASS ==========")

stock_status_summary = pd.crosstab(
    product_master["ABC_Class"],
    product_master["Stock_Status"]
)

print(stock_status_summary)
````

Step 5.4 — Create Purchasing Priority

````python
# ==========================================================
# STEP 5.4 — CREATE PURCHASING PRIORITY
# ==========================================================

def purchasing_priority(row):

    # ----------------------------------
    # NO SALES
    # ----------------------------------
    if row["ABC_Class"] == "N":
        return "Do Not Reorder / Review"

    # ----------------------------------
    # A PRODUCTS
    # ----------------------------------
    elif (
        row["ABC_Class"] == "A"
        and row["Stock_Status"] == "Out of Stock"
    ):
        return "Urgent Review"

    elif (
        row["ABC_Class"] == "A"
        and row["Stock_Status"] == "Low Coverage"
    ):
        return "High Priority"

    # ----------------------------------
    # B PRODUCTS
    # ----------------------------------
    elif (
        row["ABC_Class"] == "B"
        and row["Stock_Status"] == "Out of Stock"
    ):
        return "Reorder Review"

    elif (
        row["ABC_Class"] == "B"
        and row["Stock_Status"] == "Low Coverage"
    ):
        return "Reorder Review"

    # ----------------------------------
    # C PRODUCTS
    # ----------------------------------
    elif (
        row["ABC_Class"] == "C"
        and row["Stock_Status"] == "Out of Stock"
    ):
        return "Low Priority Review"

    # ----------------------------------
    # ENOUGH STOCK
    # ----------------------------------
    elif row["Stock_Status"] == "High Coverage":
        return "Hold / Monitor"

    else:
        return "Normal"


product_master["Purchasing_Priority"] = (
    product_master.apply(
        purchasing_priority,
        axis=1
    )
)


# ----------------------------------------------------------
# CHECK RESULT
# ----------------------------------------------------------

print("========== PURCHASING PRIORITY ==========")

priority_summary = (
    product_master["Purchasing_Priority"]
    .value_counts()
)

print(priority_summary)


print("\n========== PRIORITY BY ABC CLASS ==========")

print(
    pd.crosstab(
        product_master["ABC_Class"],
        product_master["Purchasing_Priority"]
    )
)
````

````python
# ==========================================================
# STEP 5.8 — ADD RECENT PURCHASE INFO TO PRODUCT MASTER
# ==========================================================

# ----------------------------------------------------------
# 1. SUMMARIZE CLEAN PURCHASE DATA BY PRODUCT
# ----------------------------------------------------------

purchase_summary = (
    ashley_purchase_clean
    .groupby("Stock Code")
    .agg(
        Recent_Purchase_Qty=("Quantity", "sum"),
        Last_Purchase_Date=("Date", "max")
    )
    .reset_index()
)


# ----------------------------------------------------------
# 2. MERGE INTO PRODUCT MASTER
# ----------------------------------------------------------

product_master = product_master.merge(
    purchase_summary,
    on="Stock Code",
    how="left"
)


# ----------------------------------------------------------
# 3. PRODUCTS WITHOUT PURCHASE RECORD
# ----------------------------------------------------------

product_master["Recent_Purchase_Qty"] = (
    product_master["Recent_Purchase_Qty"]
    .fillna(0)
)

product_master["Recent_Purchase_Qty"] = (
    product_master["Recent_Purchase_Qty"]
    .astype(int)
)


# ----------------------------------------------------------
# 4. CREATE RECENT PURCHASE FLAG
# ----------------------------------------------------------

product_master["Has_Recent_Purchase"] = (
    product_master["Recent_Purchase_Qty"] > 0
)


# ----------------------------------------------------------
# 5. CHECK RESULT
# ----------------------------------------------------------

print("========== RECENT PURCHASE CHECK ==========")

print(
    "Products With Recent Purchase:",
    product_master["Has_Recent_Purchase"].sum()
)

print(
    "Total Recent Purchase Units:",
    product_master["Recent_Purchase_Qty"].sum()
)

print("\nRecent Purchase Products:")

print(
    product_master[
        product_master["Has_Recent_Purchase"]
    ][
        [
            "Stock Code",
            "Description",
            "ABC_Class",
            "On Hand Qty",
            "Avg_Monthly_Sales",
            "Suggested_Reorder_Qty",
            "Recent_Purchase_Qty",
            "Last_Purchase_Date"
        ]
    ]
    .sort_values(
        "Recent_Purchase_Qty",
        ascending=False
    )
    .to_string(index=False)
)
````

Step 5.10 — Create and Save Purchase History


````python
# ==========================================================
# STEP 5.10 — SAVE CLEAN PURCHASE HISTORY
# ==========================================================

import os

PURCHASE_HISTORY_FILE = "purchase_history.csv"

# Columns used to identify the same purchase record
purchase_key = [
    "Branch",
    "Date",
    "Stock Code",
    "Quantity",
    "Buying Price",
    "Location"
]


# ----------------------------------------------------------
# 1. CHECK IF PURCHASE HISTORY ALREADY EXISTS
# ----------------------------------------------------------

if os.path.exists(PURCHASE_HISTORY_FILE):

    purchase_history = pd.read_csv(
        PURCHASE_HISTORY_FILE
    )

    purchase_history["Date"] = pd.to_datetime(
        purchase_history["Date"],
        errors="coerce"
    )

    print("Existing purchase history found.")

else:

    purchase_history = pd.DataFrame(
        columns=ashley_purchase_clean.columns
    )

    print("No existing purchase history.")
    print("Creating new purchase history.")


# ----------------------------------------------------------
# 2. NUMBER OF OLD RECORDS
# ----------------------------------------------------------

old_rows = len(purchase_history)


# ----------------------------------------------------------
# 3. COMBINE OLD HISTORY + CURRENT PURCHASE DATA
# ----------------------------------------------------------

combined_purchase_history = pd.concat(
    [
        purchase_history,
        ashley_purchase_clean
    ],
    ignore_index=True
)


# ----------------------------------------------------------
# 4. REMOVE RECORDS THAT ALREADY EXIST
# ----------------------------------------------------------

combined_purchase_history = (
    combined_purchase_history
    .drop_duplicates(
        subset=purchase_key,
        keep="first"
    )
    .copy()
)


# ----------------------------------------------------------
# 5. CALCULATE HOW MANY NEW RECORDS WERE ADDED
# ----------------------------------------------------------

new_rows = (
    len(combined_purchase_history)
    - old_rows
)


# ----------------------------------------------------------
# 6. SAVE PURCHASE HISTORY
# ----------------------------------------------------------

combined_purchase_history.to_csv(
    PURCHASE_HISTORY_FILE,
    index=False
)


# ----------------------------------------------------------
# 7. FINAL CHECK
# ----------------------------------------------------------

print("\n========== PURCHASE HISTORY SAVED ==========")

print("Previous Records:", old_rows)
print("New Records Added:", new_rows)
print(
    "Duplicate / Existing Records Skipped:",
    len(ashley_purchase_clean) - new_rows
)

print(
    "Total Purchase History Records:",
    len(combined_purchase_history)
)

print("\nIMPORTANT:")
print("Purchase history saved.")
print("Inventory On Hand Qty was NOT changed.")
````

This means the 15 purchase records are now stored separately in purchase_history.csv, exactly as we wanted.

Step 5.11 — Final Purchasing Recommendation Table

````python
# ==========================================================
# STEP 5.11 — FINAL PURCHASING RECOMMENDATION TABLE
# ==========================================================

# ----------------------------------------------------------
# 1. SELECT IMPORTANT DECISION COLUMNS
# ----------------------------------------------------------

recommendation_columns = [
    "Stock Code",
    "Description",
    "ABC_Class",
    "Price Range",
    "Selling Price",
    "Buying Price",
    "Total_Qty_Sold",
    "Total_Sales_Value",
    "Avg_Monthly_Sales",
    "On Hand Qty",
    "Stock_Coverage_Months",
    "Stock_Status",
    "Purchasing_Priority",
    "Suggested_Reorder_Qty",
    "Recent_Purchase_Qty",
    "Last_Purchase_Date"
]


purchase_recommendation = product_master[
    recommendation_columns
].copy()


# ----------------------------------------------------------
# 2. CREATE RECOMMENDATION ACTION
# ----------------------------------------------------------

def recommendation_action(row):

    # No historical sales
    if row["ABC_Class"] == "N":
        return "Review / Do Not Reorder"

    # A product + out of stock
    if (
        row["ABC_Class"] == "A"
        and row["On Hand Qty"] <= 0
    ):
        return "Urgent Reorder"

    # Product needs more stock
    if row["Suggested_Reorder_Qty"] > 0:
        return "Reorder"

    # Too much stock
    if row["Stock_Status"] == "High Coverage":
        return "Hold / Monitor"

    return "No Reorder Needed"


purchase_recommendation["Recommendation"] = (
    purchase_recommendation.apply(
        recommendation_action,
        axis=1
    )
)


# ----------------------------------------------------------
# 3. ADD RECENT PURCHASE WARNING
# ----------------------------------------------------------

purchase_recommendation["Purchase_Note"] = "No Recent Purchase"

purchase_recommendation.loc[
    purchase_recommendation["Recent_Purchase_Qty"] > 0,
    "Purchase_Note"
] = "Recent Purchase Exists"


# ----------------------------------------------------------
# 4. CREATE PRIORITY ORDER FOR SORTING
# ----------------------------------------------------------

priority_order = {
    "Urgent Reorder": 1,
    "Reorder": 2,
    "No Reorder Needed": 3,
    "Hold / Monitor": 4,
    "Review / Do Not Reorder": 5
}

purchase_recommendation["Priority_Order"] = (
    purchase_recommendation["Recommendation"]
    .map(priority_order)
)


# ----------------------------------------------------------
# 5. SORT MOST IMPORTANT PRODUCTS FIRST
# ----------------------------------------------------------

purchase_recommendation = (
    purchase_recommendation
    .sort_values(
        [
            "Priority_Order",
            "Total_Sales_Value"
        ],
        ascending=[True, False]
    )
    .reset_index(drop=True)
)


# ----------------------------------------------------------
# 6. CHECK RESULT
# ----------------------------------------------------------

print("========== FINAL PURCHASING RECOMMENDATION ==========")

print("Total Products:", len(purchase_recommendation))

print("\nRecommendation Summary:")

print(
    purchase_recommendation["Recommendation"]
    .value_counts()
)

print("\nTotal Suggested Reorder Units:")

print(
    purchase_recommendation["Suggested_Reorder_Qty"]
    .sum()
)

print("\nTop 20 Products Needing Attention:")

print(
    purchase_recommendation[
        [
            "Stock Code",
            "Description",
            "ABC_Class",
            "On Hand Qty",
            "Avg_Monthly_Sales",
            "Stock_Coverage_Months",
            "Suggested_Reorder_Qty",
            "Recommendation",
            "Purchase_Note"
        ]
    ]
    .head(20)
    .round(2)
    .to_string(index=False)
)
````
